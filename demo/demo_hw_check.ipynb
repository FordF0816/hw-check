# The autograder never reads your logic.

It compares what your program **prints**, character by character.

| what you wrote | what the grader sees | score |
|:--|:--|:--:|
| correct logic, one extra `print()` | one more line than expected | **0** |
| `round(x, 2)` instead of `f"{x:.2f}"` | `20.0` instead of `20.00` | **0** |
| `__init__` returns a value | `TypeError` | **0** |
| function name has a typo | `NameError` | **0** |

`pytest` checks logic. `flake8` checks style. **Nothing checks this.**

# `hw-check`

Ten rules, run on the **`ast`** — your code is never executed.

- the function exists, and is spelled exactly right
- a parameter is overwritten before it is used
- `print()` where the spec wants a `return`
- `input()` inside a function
- `round()` where the output needs fixed decimals
- `__init__` returns something
- an attribute is never stored on `self`
